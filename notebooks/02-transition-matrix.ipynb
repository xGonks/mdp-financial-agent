```python
import pandas as pd
import numpy as np

data = pd.read_csv("../data/external/bitcoin.csv")
data.info()
```

```python
from mdp.utils.discretize_feature import discretize_feature

data["discrete Open"] = discretize_feature(feature=data["Open"],
                                      strategy="quantile-based",
                                      ngroups=4,
                                      prefix=data["Open"].name)

data["discrete Close"] = discretize_feature(feature=data["Close"],
                                      strategy="quantile-based",
                                      ngroups=3,
                                      prefix=data["Close"].name)

data["discrete High"] = discretize_feature(feature=data["High"],
                                      strategy="quantile-based",
                                      ngroups=2,
                                      prefix=data["High"].name)

#data["discrete High"].
```

```python
from itertools import product

comb = list(product(data["discrete Open"].unique(), data["discrete Close"].unique(), data["discrete High"].unique()))

for x in comb:
    print(x)

print()
print(len(comb))
```

```python
com = list(product(data["discrete Open"].unique()))

for x in com:
    print(x)

print()
print(len(com))
```
